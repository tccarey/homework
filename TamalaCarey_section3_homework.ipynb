    - Clear the 1st set and print the result
    - Discard an element, and remove another element from the 3rd set
    - Print the remainder of the 3rd set


```python
set_1 = {'True', 'female', 29}
print(set_1)
fruit_list = ['orange', 'blueberries', 'apple']
set_1.update(fruit_list)
print(set_1)

set_1.add('toyota')
print(set_1)

set_2 = {'why', 'only one item at a time'}
set_3 = set_1.union(set_2)
set_2.pop()
print(set_2)

set_1.clear()
print(set_1)

set_3.discard('only one item at a time')
print(set_3)
```
