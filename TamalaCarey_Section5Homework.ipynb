10. Concatenate these 3 arrays into a new array named 'newArray'...
    ```python
            ([[25, 16]])
            ([[11, 2], [13, 4]])
            ([[7, 81], [5, 6], [11, 12]])
    ```

````python
import numpy as np
d = np.array([[25, 16]])
e = np.array([[11, 2], [13, 4]])
f = np.array([[7, 81], [5, 6], [11, 12]])

newArray = np.concatenate((d, e, f))
print("newArray...", newArray)
````

13. Combined and sort the following arrays into one called 'comboArray' ...

    ```python
        one = ([10, 11, 12, 13, 14, 15, 16, 17])
        two = ([20, 21, 22, 23, 24, 25, 26, 27])
        three = ([ 0, 1, 2, 3, 4, 5, 6, 7])
    ```

````python
import numpy as np
one = ([10, 11, 12, 13, 14, 15, 16, 17])
two = ([20, 21, 22, 23, 24, 25, 26, 27])
three = ([ 0, 1, 2, 3, 4, 5, 6, 7])

combopresortedArray = np.concatenate((one, two, three))
comboArray = np.sort(combopresortedArray)

print(comboArray)
````

15. Using `Series`, create a `DataFrame` that looks like this:

    | Ingredients | Quantity | Unit |
    |----|----|----|
    | Flour | 4 | cups |
    | Milk | 1 | cup |
    | Eggs | 2 | large |
    | Spam | 1 | can |

    Name: Dinner, dtype: object

````python
import pandas as pd
ingredients = pd.Series(["Flour", "Milk", "Eggs", "Spam"], name= 'ingredients')
quantity = pd.Series([4, 1, 2, 1], name= 'quantity')
unit = pd.Series(["cups", "cup", "large", "can"], name= 'unit')

df = pd.concat({'ingredients': ingredients, 
                'quantity': quantity, 
                'unit': unit}, axis=1)
print(df)
````

16. Take this data and create a DataFrame named studentData
    ```Python
        {'Name': ['Jai', 'janusha', 'Gaurav', 'Anuj'],
            'Height': [5.1, 6.2, 5.1, 5.2],
            'Qualification': ['Msc', 'MA', 'Msc', 'Msc'],
            'address': ['Delhi', 'Doha', 'Chennai', 'Dakhar'],
            'Age': [21, 23, 24, 21],
            'Pets': ['Dog', 'Bunny', 'Chinchilla', 'Parrot'],
            'sport': ['Darts', 'Basketball', 'PaddleBoarding', 'Cricket']
        }
    ```

````python
import pandas as pd
data = {'Name': ['Jai', 'janusha', 'Gaurav', 'Anuj'],
            'Height': [5.1, 6.2, 5.1, 5.2],
            'Qualification': ['Msc', 'MA', 'Msc', 'Msc'],
            'address': ['Delhi', 'Doha', 'Chennai', 'Dakhar'],
            'Age': [21, 23, 24, 21],
            'Pets': ['Dog', 'Bunny', 'Chinchilla', 'Parrot'],
            'sport': ['Darts', 'Basketball', 'PaddleBoarding', 'Cricket']
        }
studentData = pd.DataFrame(data)
print(studentData)
````

17. Add a new column to the DataFrame with the following deserts:
        ["ice cream", "Cashew Fudge", "waffels", "Carrot Halwa"]

````python
new_column = {
    "desserts": ["ice cream", "Cashew Fudge", "waffels", "Carrot Halwa"]
}
desserts = pd.DataFrame(new_column)

df = pd.DataFrame(studentData).join(desserts)
print(df)
````

18. Sort the 'studentData' DataFrame in Ascending order -- Sorting by column 'Name' and then "address"

````python
df.sort_values(by = ['Name', 'address'], inplace=True)
print(df)
````
